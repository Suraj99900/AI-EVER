# How to Change the Python Version in Google Colab

This notebook provides the step-by-step commands to switch the default Python version in a Google Colab environment from the default (Python 3.10) to **Python 3.11**.

**IMPORTANT NOTES:**
1.  You must run these commands **every time** you start a new Colab session, as the environment resets.
2.  After changing the Python version, you **must reinstall any libraries** you need (like `torch`, `transformers`, etc.) because the packages installed for the old version will not be available to the new one.

### Step 1: Check the Current Python Version

```python
# This will show the default version, likely 3.10.x
!python --version
```

### Step 2: Install Python 3.11

We will use the `deadsnakes` PPA (Personal Package Archive), which is a trusted source for newer Python versions on Ubuntu (the operating system Colab uses).

```python
# Add the deadsnakes PPA
!sudo add-apt-repository ppa:deadsnakes/ppa -y
!sudo apt-get update

# Install Python 3.11 and its development tools
!sudo apt-get install python3.11 python3.11-dev python3.11-distutils -y
```

### Step 3: Set Python 3.11 as the Default

We use the `update-alternatives` command to tell the system that `python3` should now point to our newly installed Python 3.11.

```python
# Make python3.11 the default python3
!sudo update-alternatives --install /usr/bin/python3 python3 /usr/bin/python3.10 1
!sudo update-alternatives --install /usr/bin/python3 python3 /usr/bin/python3.11 2

# Choose python3.11 as the default
!sudo update-alternatives --config python3
```

*(When you run the cell above, it might ask you to choose a version. Enter the number corresponding to `python3.11` and press Enter.)*

### Step 4: Verify the New Version

```python
# This should now output "Python 3.11.x"
!python --version
```

### Step 5: Install `pip` for Python 3.11

The default `pip` is linked to the old Python version. We need to install it for our new version.

```python
# Install pip for Python 3.11
!curl -sS https://bootstrap.pypa.io/get-pip.py | python3.11

# Verify pip version
!pip --version
```

### Step 6: Reinstall Your Project Dependencies

Your environment is now running Python 3.11. You can now install your project's packages. **This is a required step!**

```python
# Example: Reinstall the libraries from your requirements file
!pip install -q accelerate peft bitsandbytes transformers datasets torch

print("✅ Environment is now running Python 3.11 with required packages installed.")
```
